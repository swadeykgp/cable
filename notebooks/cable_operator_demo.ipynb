# CABLE operator demo
CPU-only demonstration; random inputs, no training or performance results.


```python
import torch
from torch import nn

class CABLE2D(nn.Module):
    def __init__(self, channels=32, groups=4):
        super().__init__()
        assert channels % groups == 0
        self.local_mix = nn.Conv2d(channels, channels, 3, padding=1, groups=channels, bias=False)
        self.head_proj = nn.Conv2d(channels, channels, 1, groups=groups, bias=False)
        self.out_proj = nn.Conv2d(channels, channels, 1, bias=False)
    def forward(self, x):
        return x + self.out_proj(self.head_proj(self.local_mix(x)))
```

```python
block = CABLE2D(32, 4)
x = torch.randn(1, 32, 64, 64)
y = block(x)
print("Input:", tuple(x.shape), "Output:", tuple(y.shape))
print("Operator weights:", sum(p.numel() for p in block.parameters()))
```
